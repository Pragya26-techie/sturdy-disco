```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int,b:int):
    """
    it will return sum of two numbers.
    Args:
    a:num 1,b:num 2
    """
    return a+b

@tool
def multiply_numbers(a:int,b:int):
    """
    it will return product of two numbers.
    Args:
    a:num 1,b:num 2
    """
    return a*b
```

```python
add_numbers.invoke({"a":12,"b":3})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
```

```python
## for creating agent
from langchain.agents import create_agent
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers,multiply_numbers],
    system_prompt="You are a math teacher,and always use tool for calculation."
)
```

```python
response = agent.invoke({"messages":[{"role":"user","content":"what is 2 + 30 * 10"}]})
print(response["messages"][-1].content)
# for message in response["messages"]:
#     print("ROLE:", message.type)
#     print("CONTENT:", message.content)
#     print("-" * 50)
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[add_numbers, multiply_numbers],
    system_prompt="""
    You are a math teacher.
    Always use the available tools for calculations.

    Follow standard mathematical operator precedence.
    Perform multiplication before addition.
    Break complex expressions into the correct sequence of operations.
    """
)
```

```python
response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "what is 2 + 30 * 10"
        }
    ]
})

print(response["messages"][-1].content)
```
