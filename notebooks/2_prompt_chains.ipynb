```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

* standard format to pass prompts to llm is to pass it in dictionary format.

```python
prompts = [
    {"role":"system","content":"You are a JavaScript Developer ?"},
    {"role":"user","content":"Difference b/w map and filter"},
]
res = llm.invoke(prompts)
print(res.content)
```

#### Prompt Templates - Dynamic Prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
llm = ChatGoogleGenerativeAI(model = "gemini-3.5-flash")
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role":"system","content":"You are a translator and translate input in {language} ?"},
    {"role":"user","content":"{query}"},
])
prompts
```

```python
prompts.invoke()
```

```python
res = llm.invoke(prompts.invoke({"language":"Hindi","query":"I love you"}))
res.content
```

#### Chains - Runnable

```python
chains = prompts | llm
```

```python
chains
```

```python
res = chains.invoke({"language":"Hindi","query":"I love langchain"})
print(res.content)
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
out = StrOutputParser()
```

```python
chains = prompts | llm|out
```

```python
res = chains.invoke({"language":"Hindi","query":"I love langchain"})
print(res)
```
