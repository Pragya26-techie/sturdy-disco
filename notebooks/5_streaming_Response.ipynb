### Streaming Response

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b",streaming=True)
```

```python
question = "what is genai"
```

```python
response = llm.stream(question)
```

```python
## generator concept
response = llm.stream(question)
for chunk in response:
    print(chunk.content,end="")
```
