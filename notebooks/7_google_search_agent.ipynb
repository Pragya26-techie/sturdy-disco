```python
from dotenv import load_dotenv
load_dotenv()
```

```python
## creating llm+search tool
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b")
search = GoogleSerperAPIWrapper()
```

```python
search.run("Address of TMBU")
```

```python
## Connecting Agent with llm
agent = create_agent(
    model=llm,
    tools=[search.run],
    system_prompt="You are a agent and can search for any question on google."
)
```

```python
question="what are pros and cons of becomig IFS offficer"
```

```python
response = agent.invoke({"messages":[{"role":"user","content":question}]})
```

```python
len(response["messages"])
```

```python
print(response["messages"][-1].content)
```

```python
from langgraph.checkpoint.memory import InMemorySaver
```

```python
agent = create_agent(
    model=llm,
    tools=[search.run],
    system_prompt="You are a agent and can search for any question on google.",
    checkpointer = InMemorySaver()
)
```

```python
question1="who won the woman's world cup 2025"
```

```python
response = agent.invoke({"messages":[{"role":"user","content":question1}]},
{"configurable":{"thread_id":"pragya"}},)
```

```python
print(response["messages"][-1].content)
```

```python
question2="Total Runsn?"
```

```python
response = agent.invoke({"messages":[{"role":"user","content":question2}]},
{"configurable":{"thread_id":"pragya"}},)
```

```python
print(response["messages"][-1].content)
```
